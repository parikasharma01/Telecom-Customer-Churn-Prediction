```python
import matplotlib.pyplot as plt

df['Churn'].value_counts().plot(kind='bar')
plt.title('Customer Churn Distribution')
plt.xlabel('Churn')
plt.ylabel('Number of Customers')
plt.show()
```

```python
contract_churn = pd.crosstab(
    df['Contract'],
    df['Churn'],
    normalize='index'
) * 100

contract_churn.plot(kind='bar', figsize=(8,5))

plt.title('Churn Rate by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Churn Percentage (%)')
plt.xticks(rotation=0)
plt.legend(title='Churn')
plt.tight_layout()
plt.show()
```

```python
internet_churn = pd.crosstab(
    df['InternetService'],
    df['Churn'],
    normalize='index'
) * 100

internet_churn.plot(kind='bar', figsize=(8,5))

plt.title('Churn Rate by Internet Service')
plt.xlabel('Internet Service')
plt.ylabel('Churn Percentage (%)')
plt.xticks(rotation=0)
plt.legend(title='Churn')
plt.tight_layout()
plt.show()
```

```python
payment_churn = pd.crosstab(
    df['PaymentMethod'],
    df['Churn'],
    normalize='index'
) * 100

payment_churn.plot(kind='bar', figsize=(9,5))

plt.title('Churn Rate by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Churn Percentage (%)')
plt.xticks(rotation=20)
plt.legend(title='Churn')
plt.tight_layout()
plt.show()
```

```python
tenure_churn = df.groupby('Churn')['tenure'].mean()

tenure_churn.plot(kind='bar', figsize=(7,5))

plt.title('Average Tenure by Churn Status')
plt.xlabel('Churn')
plt.ylabel('Average Tenure (Months)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python


df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

df = pd.get_dummies(df, drop_first=True)

df.head()
```

```python
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt

X = df.drop('Churn', axis=1)
y = df['Churn']

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

importance = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

importance.head(10).plot(kind='bar', figsize=(8,5))

plt.title('Top 10 Feature Importance')
plt.xlabel('Features')
plt.ylabel('Importance')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
